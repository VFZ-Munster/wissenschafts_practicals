```python
import math

# Continuous variables to summarize. Edit this list if you want more/fewer rows.
variables = [
    "nteeth",
    "cal_mean",
    "bop",
    "age",
    "height",
    "weight",
    "bmi",
    "wc",
    "ln_loss",
]

def mean_ci(series):
    x = pd.to_numeric(series, errors="coerce").dropna()
    n = len(x)
    if n == 0:
        return "NA"
    mean = x.mean()
    if n == 1:
        return f"{mean:.2f} (NA, NA)"
    se = x.std(ddof=1) / math.sqrt(n)
    lower = mean - 1.96 * se
    upper = mean + 1.96 * se
    return f"{mean:.2f} ({lower:.2f}, {upper:.2f})"

rows = []
for var in variables:
    rows.append({
        "Variable": var,
        "Male": mean_ci(df.loc[df["sex"] == 2, var]),
        "Female": mean_ci(df.loc[df["sex"] == 1, var]),
        "Total": mean_ci(df[var]),
    })

summary_table = pd.DataFrame(rows)
summary_table
```
